```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
df = pd.read_csv(r"C:\Users\Abhay\Downloads\Multiple_Regression_HousePrice_Dataset.csv")
print(df.head(10))
print(df.shape)  
df.info()
X = df[["Area_sqft", "Bedrooms", "Age_years", "Distance_City_km"]]
y = df["House_Price_Lakhs"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
model = LinearRegression()
model.fit(X_train, y_train)
print("Intercept:", model.intercept_)
coefficient_table = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_})
print(coefficient_table)
y_pred = model.predict(X_test)
results = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred})
print(results.head(10))
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
plt.scatter(y_test, y_pred)
minimum = min(y_test.min(), y_pred.min())
maximum = max(y_test.max(), y_pred.max())
plt.plot([minimum, maximum], [minimum, maximum], linestyle="--")
plt.xlabel("Actual House Price (Lakhs)")
plt.ylabel("Predicted House Price (Lakhs)")
plt.title("Actual Price vs Predicted Price")
plt.show()
new_house = pd.DataFrame({
    "Area_sqft": [1800],
    "Bedrooms": [3],
    "Age_years": [5],
    "Distance_City_km": [7]})
predicted_price = model.predict(new_house)
print("Predicted Price:", predicted_price[0], "Lakhs")
```
